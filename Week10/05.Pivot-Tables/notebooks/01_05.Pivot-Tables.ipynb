## Overview
A **Pivot Table** is an essential data analysis tool that reshapes flat tabular data into a two-dimensional summary matrix. While `.groupby()` primarily creates tall hierarchical listings, `pd.pivot_table()` arranges one categorical variable along the rows and another along the columns, making cross-category comparison effortless.

In this notebook, we cover:
1. **`pd.pivot_table()` Parameters**: `index`, `columns`, `values`, and `aggfunc`.
2. **GroupBy vs. Pivot Tables**: When to choose a flat grouping versus a 2D matrix.
3. **Handling Sparse Combinations**: Replacing `NaN` using `fill_value=0`.
4. **Multi-Metric Summaries**: Combining multiple statistical functions across dimensions.

## 1. Setup: Tech Graduate Salaries Dataset

We analyse starting salaries (kAUD) and placement rates for Australian university IT graduates across states, degrees, and specialisations.

```python
import pandas as pd
import numpy as np

df_tech = pd.DataFrame({
    'state': [
        'NSW', 'NSW', 'NSW', 'NSW',
        'VIC', 'VIC', 'VIC', 'VIC',
        'QLD', 'QLD', 'QLD', 'QLD'
    ],
    'degree': [
        'Bachelor', 'Master', 'Bachelor', 'Master',
        'Bachelor', 'Master', 'Bachelor', 'Master',
        'Bachelor', 'Master', 'Bachelor', 'Master'
    ],
    'specialisation': [
        'Software Eng', 'Software Eng', 'Cyber Security', 'Data Science',
        'Software Eng', 'Cyber Security', 'Data Science', 'Data Science',
        'Software Eng', 'Cyber Security', 'Data Science', 'Cloud Computing'
    ],
    'salary_kaud': [85, 105, 90, 110, 82, 98, 92, 108, 78, 88, 85, 102],
    'placement_pct': [92, 96, 94, 95, 90, 93, 91, 97, 88, 91, 89, 94]
})

print("Tech Employment Dataset:")
display(df_tech)
```

## 2. Basic 2D Pivot Table

A pivot table requires:
- `index`: Column to use as row headers.
- `columns`: Column to use as column headers.
- `values`: Numeric column to aggregate.
- `aggfunc`: Function to apply (default is `'mean'`).
- `fill_value`: Value to substitute for missing coordinates.

```python
salary_table = pd.pivot_table(
    df_tech,
    values='salary_kaud',
    index='state',
    columns='specialisation',
    aggfunc='mean',
    fill_value=0
)

print("Average Salary (kAUD) by State and Specialisation:")
display(salary_table.round(1))
```

## 3. Contrasting GroupBy vs. Pivot Table

- `.groupby()` produces a 1D Series with a MultiIndex.
- `pd.pivot_table()` produces a 2D matrix directly, equivalent to `.groupby().unstack()`.

```python
print("1. GroupBy Output (Hierarchical Series):")
display(df_tech.groupby(['state', 'specialisation'])['salary_kaud'].mean())

print("\n2. Pivot Table Output (2D Matrix):")
display(pd.pivot_table(df_tech, values='salary_kaud', index='state', columns='specialisation', aggfunc='mean'))
```

## 4. Multi-Level and Multi-Function Pivot Tables

You can specify multiple row indexes (e.g. `index=['state', 'degree']`) and multiple aggregation functions (`aggfunc=['mean', 'count']`).

```python
multi_pivot = pd.pivot_table(
    df_tech,
    values='salary_kaud',
    index=['state', 'degree'],
    columns='specialisation',
    aggfunc=['mean', 'count'],
    fill_value=0
)

print("Multi-Level Pivot Table (Mean and Count):")
display(multi_pivot)
```

## 6. Key Takeaways

1. **`pd.pivot_table()`**: The preferred tool for 2D cross-tabulation of continuous measures against two categorical dimensions.
2. **Default Function**: `aggfunc` defaults to `'mean'`. Specify `'sum'`, `'count'`, etc., as needed.
3. **`fill_value`**: Replaces `NaN` cleanly when combinations do not exist in the data.
4. **Equivalence**: A basic pivot table is functionally equivalent to `df.groupby(['A', 'B'])['C'].mean().unstack()`.
