## Overview
In real-world data science, data frequently arrives in formats beyond CSV:
- **Microsoft Excel (`.xlsx`)**: Ubiquitous in business reporting, featuring multiple sheets, formatting, and formulas.
- **JSON (JavaScript Object Notation)**: The universal standard for REST APIs, cloud databases, and web services.

In this notebook, you will learn how to read data from Excel workbooks (`energy_stats.xlsx`) and JSON files (`sydney_airbnb.json`) into Pandas DataFrames.

## 1. Reading JSON Files with pd.read_json()
When a JSON file consists of an array of records (list of dictionaries), `pd.read_json()` parses it directly into rows and columns.

````python
import pandas as pd

# Read Sydney Airbnb JSON file
df_airbnb = pd.read_json('sydney_airbnb.json')
display(df_airbnb)

print(f"Total listings: {len(df_airbnb)}")
print(f"Average price: ${df_airbnb['price'].mean():.2f} AUD")
print(f"Average rating: {df_airbnb['rating'].mean():.2f} / 5.0")
````

## 2. Flattening Nested JSON with pd.json_normalize()
Web APIs often return hierarchical data where dictionaries contain sub-dictionaries. Use `pd.json_normalize()` to flatten nested structures into dot-notated columns.

````python
nested_api_response = [
    {
        'student_id': 'S101',
        'name': 'Nguyen An',
        'campus': {'city': 'Sydney', 'state': 'NSW'},
        'enrolment': {'course': 'ITEC102', 'semester': 1}
    },
    {
        'student_id': 'S102',
        'name': 'Sarah Miller',
        'campus': {'city': 'Melbourne', 'state': 'VIC'},
        'enrolment': {'course': 'ITEC102', 'semester': 1}
    }
]

df_flat = pd.json_normalize(nested_api_response)
display(df_flat)
````

## 3. Reading Excel Worksheets with pd.read_excel()
Use `pd.read_excel()` to load tabular spreadsheets. The `sheet_name` parameter lets you select which worksheet tab to load.

````python
try:
    df_energy = pd.read_excel('energy_stats.xlsx', sheet_name='NSW')
    display(df_energy)
    print(f"\nTotal NSW Clean Energy Produced: {df_energy['Energy_Produced'].sum()} GWh")
except ImportError:
    print("Install openpyxl via 'uv add openpyxl' or 'pip install openpyxl' to read Excel files.")
````

## 4. Discovering Excel Sheets with pd.ExcelFile
To see all sheets present in a workbook before loading:

````python
try:
    xl = pd.ExcelFile('energy_stats.xlsx')
    print("Available sheets:", xl.sheet_names)
except ImportError:
    print("openpyxl required.")
````

## Enrichment
### Format Trade-offs: CSV vs Excel vs JSON
| Feature | CSV | Excel (`.xlsx`) | JSON |
|---|---|---|---|
| **Primary Use** | Universal data exchange | Business spreadsheets, multi-tab reports | Web APIs, semi-structured logs |
| **Speed & Size** | Fast, compact plain text | Slower, heavier XML-based zip | Flexible, slightly verbose text |
| **Hierarchy** | Flat tables only | Multiple flat 2D sheets | Nested dictionaries & arrays |
| **Pandas Function** | `pd.read_csv()` | `pd.read_excel()` | `pd.read_json()` / `json_normalize()` |

## Takeaways
- Use `pd.read_json()` to load flat JSON records directly into DataFrames.
- Use `pd.json_normalize()` to flatten nested, hierarchical JSON responses from web APIs.
- Use `pd.read_excel()` with `sheet_name='SheetName'` to load specific tabs from Excel workbooks.
- Use `pd.ExcelFile(file_path).sheet_names` to discover what sheets exist before loading.
- Ensure `openpyxl` is installed in your Python environment when working with `.xlsx` files.

## Conclusion
Real-world data analysts must be fluent across formats. Combining `pd.read_excel()`, `pd.read_json()`, and `pd.json_normalize()` ensures you can ingest whatever data format your organization or API provides.
