## Overview
Once data is loaded into a Pandas DataFrame, the immediate next step is **exploratory data analysis (EDA)**.
In this notebook, you will learn how to inspect dimensions, preview rows, check column data types and missing values, and calculate key descriptive statistics on Australian weather observations.

## 1. Loading the Dataset
Let's load `sample_weather.csv` with dates parsed.

````python
import pandas as pd

df = pd.read_csv('sample_weather.csv', parse_dates=['Date'])
print(f"Loaded {len(df)} rows")
display(df.head())
````

## 2. Structural Inspection: head(), tail(), shape, columns
Inspect the beginning, ending, and dimensions of the DataFrame.

````python
print('First 3 rows:')
display(df.head(3))

print('Last 2 rows:')
display(df.tail(2))

print(f'Dimensions (rows, columns): {df.shape}')
print(f'Column names: {list(df.columns)}')
````

## 3. Systemic Metadata: info() and dtypes
`df.info()` gives an immediate view of column non-null counts, types, and memory consumption.

````python
df.info()
````

## 4. Descriptive Statistics: describe()
`df.describe()` calculates summary statistics (mean, std, min, percentiles, max) for numerical fields.

````python
print('Summary of numerical columns:')
display(df.describe())

print('Summary including categorical columns:')
display(df.describe(include='all'))
````

## 5. Targeted Aggregations & Frequency Counts
Compute specific summary metrics and explore categorical frequency distributions.

````python
print(f"Mean Temperature: {df['Temperature'].mean():.2f}°C")
print(f"Median Temperature: {df['Temperature'].median():.2f}°C")
print(f"Total Rainfall: {df['Rainfall'].sum():.1f} mm")

print("\nObservation counts per City:")
display(df['City'].value_counts())

print(f"Unique cities ({df['City'].nunique()}): {list(df['City'].unique())}")
````

## Enrichment
### Finding Records at Extremes (.idxmax())
`.idxmax()` retrieves the index of the highest value, allowing you to fetch the entire matching record:
```python
hottest_idx = df['Temperature'].idxmax()
print(df.loc[hottest_idx])
```

### Grouped Summaries with groupby()
Group data by city to inspect localized climate statistics:
```python
print(df.groupby('City')['Temperature'].mean())
```

## Takeaways
- Always start exploratory data analysis with `df.info()` to verify data types and detect missing values.
- Use `df.describe()` to quickly gauge the spread, mean, and range of numerical features.
- Use `.value_counts()` to inspect the representation and balance of categories.
- Combine `.idxmax()` with `.loc[]` to look up complete records associated with minimum or maximum values.

## Conclusion
Exploring and summarising data gives you vital visibility into data quality, outliers, and distributions before performing complex analytics or machine learning.
