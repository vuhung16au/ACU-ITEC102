## Overview
This notebook covers how to import tabular data stored in Comma-Separated Values (`.csv`) format into a Pandas DataFrame using `pd.read_csv()`.

CSV is the most common plain-text format used for exchanging data in business and data science. In this notebook, we analyze Australian weather observations recorded across Sydney, Melbourne, Brisbane, Perth, and Adelaide.

## 1. Basic CSV Ingestion
By default, `pd.read_csv()` assumes the first row contains column headers and fields are separated by commas.

````python
import pandas as pd

# Load the CSV file
df = pd.read_csv('sample_weather.csv')
print('First 5 rows:')
display(df.head())

print('\nColumn data types:')
print(df.dtypes)
````

## 2. Parsing Dates
Notice above that `Date` was loaded as an `object` (string). With `parse_dates=['Date']`, Pandas converts it directly to `datetime64`, unlocking temporal analysis methods.

````python
df_dates = pd.read_csv('sample_weather.csv', parse_dates=['Date'])
print(df_dates.dtypes)
print(f"\nType of first Date entry: {type(df_dates['Date'].iloc[0])}")
````

## 3. Setting an Index Column
You can designate a column (such as `Date`) as the DataFrame index during ingestion using `index_col`.

````python
df_indexed = pd.read_csv('sample_weather.csv', parse_dates=['Date'], index_col='Date')
display(df_indexed.head(3))
print('\nIndex details:')
print(df_indexed.index)
````

## 4. Memory Optimisation: usecols and nrows
When loading massive datasets, use `usecols` to load only necessary columns, and `nrows` to preview the first few rows without exhausting memory.

````python
df_subset = pd.read_csv('sample_weather.csv', usecols=['City', 'Temperature'], nrows=4)
display(df_subset)
````

## Enrichment
### The Trap of Compression
Saving tiny CSV files in compressed format (`.gz`) can actually increase their file size due to metadata overhead! Compression is meant for large files.

### Encoding Nightmares (UTF-8 vs Latin-1)
If you see unexpected symbols instead of accented characters, specify the encoding when reading a file:
```python
df = pd.read_csv('australian_data.csv', encoding='utf-8')
```

### Reading CSV directly from a URL
`pd.read_csv()` can also stream directly over HTTP from open data repositories like `data.gov.au`.

## Takeaways
- `pd.read_csv()` is the primary function to load CSV files into Pandas DataFrames.
- Use `parse_dates=['Date']` to automatically parse dates into `datetime64`.
- Use `index_col='col_name'` to set an index column upon reading.
- Use `usecols` to select specific columns and `nrows` to limit row count for memory efficiency.
- Check `df.dtypes` immediately after loading to confirm data types are correct.

## Conclusion
Reading CSV files correctly with appropriate parameters ensures your data pipeline starts with clean, correctly typed DataFrames, avoiding errors in later analysis steps.
