# Filtering Data using NumPy Masks (Boolean Arrays)

**Overview:** This notebook demonstrates how to filter data using boolean masks.

## Code Snippets
The following example demonstrates the core concept of this topic:

```python
import numpy as np

ages = np.array([15, 22, 14, 30, 18, 12])
adult_mask = ages >= 18
adults = ages[adult_mask]
print(f"Adult ages only: {adults}")
```

## Enrichment
**Broadcasting:** NumPy conceptually stretches scalars to match the size of arrays before performing mathematical operations. This occurs in C-code, providing a huge performance boost over standard Python loops.
**Upcasting:** If you try to place different types (like `int` and `float`) in a single array, NumPy will implicitly convert them to the most complex type to avoid data loss.
