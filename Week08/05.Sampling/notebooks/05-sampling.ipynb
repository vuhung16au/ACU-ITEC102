## Overview
In modern data science, datasets often contain millions or billions of rows. Running complex exploratory analyses or prototyping algorithms on an entire massive dataset can be prohibitively slow.

**Sampling** allows you to extract a representative subset of your data to speed up development, perform quality control audits, or create machine learning training/testing splits.

In this notebook, you will learn how to:
- Draw random samples by count (`n`) and fraction (`frac`).
- Guarantee scientific reproducibility using `random_state`.
- Sample with and without replacement (`replace=True/False`).
- Perform **stratified sampling** to maintain balanced demographic representation.

## 1. Load Dataset
Let's load `messy_student_data.csv`.

````python
import pandas as pd

df = pd.read_csv('messy_student_data.csv')
print(f"Loaded {len(df)} student records")
display(df.head())
````

## 2. Simple Random Sampling by Count and Fraction
Use `df.sample(n=k)` to draw `k` rows, or `df.sample(frac=p)` to draw a fraction `p` (e.g. 0.15 for 15%).

````python
# Sample exactly 5 random students
sample_5 = df.sample(n=5, random_state=42)
print("Sample of 5 students:")
display(sample_5[['StudentID', 'Name', 'Score', 'Status']])

# Sample 10% of the dataset
sample_10pct = df.sample(frac=0.10, random_state=42)
print(f"10% sample size: {len(sample_10pct)} rows")
````

## 3. Seed Reproducibility with random_state
Setting `random_state` makes your sampling deterministic, ensuring identical results across runs and collaborating teammates.

````python
run_a = df.sample(n=3, random_state=102)
run_b = df.sample(n=3, random_state=102)

print("Run A IDs:", list(run_a['StudentID']))
print("Run B IDs:", list(run_b['StudentID']))
print("Are both samples identical?", run_a.equals(run_b))
````

## 4. Sampling With vs Without Replacement
`replace=False` (default) ensures each row can only be picked once. Setting `replace=True` allows rows to be selected multiple times (used in statistical bootstrapping).

````python
sample_boot = df.sample(n=8, replace=True, random_state=7)
print(f"Sample rows: {len(sample_boot)}, Unique IDs: {sample_boot['StudentID'].nunique()}")
display(sample_boot.head(4))
````

## 5. Stratified Sampling with groupby().sample()
When a dataset contains unequal subgroups, purely random sampling may under-represent minority classes.
**Stratified sampling** guarantees an equal or proportional sample is drawn from every subgroup.

````python
# Draw 5 students from each Status category (Pass and Fail)
stratified = df.groupby('Status').sample(n=5, random_state=42)
display(stratified[['StudentID', 'Name', 'Score', 'Status']])

print("Counts per group:")
display(stratified['Status'].value_counts())
````

## Enrichment
### Creating Train / Test Splits
```python
train_df = df.sample(frac=0.80, random_state=42)
test_df = df.drop(train_df.index)
print(f"Train size: {len(train_df)}, Test size: {len(test_df)}")
```

## Takeaways
- Use `df.sample(n=...)` for exact row counts and `df.sample(frac=...)` for proportions.
- Always set `random_state` to ensure reproducible results across team members and CI pipelines.
- Use `replace=False` for standard subsets; use `replace=True` for statistical bootstrapping.
- Use `df.groupby('category').sample(...)` for stratified sampling to avoid demographic or class representation bias.

## Conclusion
Sampling allows you to rapidly iterate, validate pipelines, and prototype models without waiting on long runtimes for massive datasets.
