## Overview
In data science and software engineering, **raw data is rarely clean**. Industry surveys report that data professionals spend up to 80% of their time cleaning and preprocessing data.

Common data quality problems include:
- Missing values (`NaN`, empty strings).
- Duplicate records.
- Inconsistent or improper data types.
- Out-of-range or invalid values (e.g., negative test scores).
- Untrimmed whitespace in text fields.

In this notebook, you will learn the foundational workflow for auditing and cleaning a student performance dataset (`messy_student_data.csv`).

## 1. Auditing Dataset Health: shape, info(), and isna()
Always begin by inspecting dimensions, types, and the extent of missing values.

````python
import pandas as pd

df = pd.read_csv('messy_student_data.csv')
print(f"Shape: {df.shape} (rows, columns)")
display(df.head())

print("\nMissing values per column:")
display(df.isna().sum())

print("\nMissing percentages:")
display((df.isna().mean() * 100).round(1).astype(str) + '%')
````

## 2. Detecting and Removing Duplicate Rows
Duplicates distort sums, averages, and statistical models. Check with `.duplicated()` and remove with `.drop_duplicates()`.

````python
# Simulate a duplicate row
df_with_dupes = pd.concat([df, df.iloc[[0]]], ignore_index=True)
print(f"Total rows with dupe: {len(df_with_dupes)}")
print(f"Duplicates found: {df_with_dupes.duplicated().sum()}")

# Drop duplicates
df_clean = df_with_dupes.drop_duplicates()
print(f"Rows after .drop_duplicates(): {len(df_clean)}")
````

## 3. Data Type Consistency and Nullable Integers
In Pandas, integers containing missing values historically had to be stored as `float64`. Modern Pandas supports nullable integer types like `'Int64'` (capital 'I').

````python
print("Original column types:")
print(df.dtypes)

# Convert Age to nullable Int64
df_typed = df.copy()
df_typed['Age'] = df_typed['Age'].astype('Int64')
display(df_typed[['StudentID', 'Name', 'Age']].head(6))
````

## 4. Validating Numerical Bounds and Domain Rules
Check for invalid values (e.g. Scores outside 0-100).

````python
print(f"Score minimum: {df['Score'].min()}, maximum: {df['Score'].max()}")
invalid = df[(df['Score'] < 0) | (df['Score'] > 100)]
print(f"Total invalid scores detected: {len(invalid)}")
````

## 5. String Sanitisation with .str Accessors
Use `.str.strip()` to remove leading/trailing whitespace and `.str.title()` to standardize name casing.

````python
sample_names = pd.Series(['  nguyen van ', 'SARAH MILLER  ', ' chloe smith '])
cleaned = sample_names.str.strip().str.title()
display(pd.DataFrame({'Raw': sample_names, 'Cleaned': cleaned}))
````

## Enrichment

```python
complete_rows = df.dropna().shape[0]
completeness_pct = (complete_rows / len(df)) * 100

```

## Takeaways
- Never analyze raw data without an initial audit (`.shape`, `.info()`, `.isna().sum()`).
- Check for duplicate observations using `.duplicated()` and remove them with `.drop_duplicates()`.
- Use nullable types like `'Int64'` when integer columns contain missing values.
- Validate domain rules (e.g. check score boundaries) to detect data entry errors.
- Clean string columns with `.str.strip()` and `.str.title()` / `.str.lower()`.

## Conclusion
Data cleaning is an iterative, disciplined process. Clean data forms the reliable bedrock for all subsequent statistical analysis and predictive modeling.
