## Overview
In data analysis, continuous numerical values (such as test marks from 0 to 100 or ages from 18 to 65) are often easier to interpret and analyze when converted into discrete categorical groups or buckets. This process is called **binning** or **discretisation**.

Common applications:
- Converting continuous test scores into academic grades (Fail, Pass, Credit, Distinction, High Distinction).
- Grouping customer ages into demographic brackets (18-24, 25-34, 35-49).
- Segmenting income into quartiles.

In this notebook, you will learn how to use `pd.cut()` for custom and equal-width intervals, and `pd.qcut()` for quantile-based equal-frequency bins.

## 1. Load Dataset & Pre-Clean Missing Values
Let's load `messy_student_data.csv` and impute missing values for clean binning.

```python
import pandas as pd

df = pd.read_csv('messy_student_data.csv')
df['Score'] = df['Score'].fillna(df['Score'].mean())
df['Age'] = df['Age'].fillna(df['Age'].median())
print(f"Loaded and pre-cleaned {len(df)} records")
display(df.head())
```

## 2. Custom Bin Edges: Australian Higher Education Grades
Use `pd.cut()` with explicit interval boundaries and human-readable grade labels:
- **Fail (N)**: 0 to < 50
- **Pass (P)**: 50 to < 65
- **Credit (C)**: 65 to < 75
- **Distinction (D)**: 75 to < 85
- **High Distinction (HD)**: 85 to 100

```python
grade_bins = [0, 50, 65, 75, 85, 100]
grade_labels = ['Fail (N)', 'Pass (P)', 'Credit (C)', 'Distinction (D)', 'High Distinction (HD)']

df['Grade'] = pd.cut(df['Score'], bins=grade_bins, labels=grade_labels, right=False)
display(df[['StudentID', 'Name', 'Score', 'Grade']].head(8))

print("Grade Frequency Distribution:")
display(df['Grade'].value_counts().sort_index())
```

## 3. Automatic Equal-Width Binning
Passing an integer `bins=k` to `pd.cut()` automatically divides the range into `k` equal-width numerical intervals.

```python
df['Age_Bracket'] = pd.cut(df['Age'], bins=3)
print("Equal-width age brackets:")
display(df['Age_Bracket'].value_counts().sort_index())
```

## 4. Equal-Frequency Binning with pd.qcut() (Quantiles)
While `pd.cut()` creates bins with equal numerical span, `pd.qcut()` creates bins with **equal numbers of observations** based on percentiles (e.g., quartiles).

```python
quartile_labels = ['Q1 (Lowest 25%)', 'Q2 (Lower-Mid)', 'Q3 (Upper-Mid)', 'Q4 (Top 25%)']
df['Score_Quartile'] = pd.qcut(df['Score'], q=4, labels=quartile_labels)

print("Observations per quartile (notice balanced frequencies):")
display(df['Score_Quartile'].value_counts().sort_index())
```

## 5. Cross-Tabulation Analysis
Examine relationships across categories using `pd.crosstab()`.

```python
print("Cross-Tabulation: Grade vs Status:")
display(pd.crosstab(df['Grade'], df['Status']))
```

## Enrichment
### Left-Open vs Right-Closed Intervals
By default, `pd.cut()` uses `(a, b]` (left-open, right-closed). Pass `right=False` to produce `[a, b)` (left-closed, right-open), which matches standard inclusive grading criteria.

## Takeaways
- Binning simplifies continuous variance into actionable categorical levels.
- Use `pd.cut()` when you have specific domain thresholds (e.g. university grade cutoffs).
- Use `pd.qcut()` when you want balanced cohorts with an equal count of observations in each bucket.
- Combine binned columns with `.value_counts()` and `pd.crosstab()` to analyze demographic breakdowns.

## Conclusion
Discretising continuous metrics enables clear executive reporting, simplifies complex data patterns, and prepares features for classification models.
