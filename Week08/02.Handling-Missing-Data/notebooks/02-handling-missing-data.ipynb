## Overview
Missing data is ubiquitous in data science. Before feeding data into models or reports, analysts must decide whether to **drop** incomplete records or **impute** them.

In this notebook, you will learn how to detect missing values, inspect affected rows, and apply various dropping strategies using `df.dropna()`. You will also evaluate the trade-offs and risks of data loss.

## 1. Load Dataset & Audit Missing Entries
Let's load `messy_student_data.csv` and inspect which columns contain missing values.

````python
import pandas as pd

df = pd.read_csv('messy_student_data.csv')
print(f"Loaded {len(df)} records")
print("Missing counts:")
display(df.isna().sum())
````

## 2. Inspecting Incomplete Rows
Use `.isna().any(axis=1)` to view all records that contain at least one missing entry.

````python
incomplete_rows = df[df.isna().any(axis=1)]
print(f"Total incomplete rows: {len(incomplete_rows)}")
display(incomplete_rows.head(6))
````

## 3. Strategy A: Default Dropping (how='any')
`df.dropna()` deletes any row that contains at least one null value. While quick, this can lead to substantial data loss.

````python
df_clean_any = df.dropna()
loss_pct = ((len(df) - len(df_clean_any)) / len(df)) * 100
print(f"Original rows: {len(df)}, Retained: {len(df_clean_any)}, Data Loss: {loss_pct:.1f}%")
display(df_clean_any.head())
````

## 4. Strategy B: Targeted Dropping with subset=[...]
Rather than discarding rows missing non-essential fields (like `Age`), only drop rows where a critical field (like `Score`) is missing.

````python
df_valid_scores = df.dropna(subset=['Score'])
print(f"Rows with valid test scores: {len(df_valid_scores)}")
print(f"Missing ages still retained: {df_valid_scores['Age'].isna().sum()}")
````

## 5. Strategy C: Threshold Dropping with thresh=k
The `thresh` argument lets you specify a minimum number of non-null values required to keep a row.

````python
# Retain rows with at least 4 non-null values out of 5 columns
df_thresh = df.dropna(thresh=4)
print(f"Rows with at least 4 non-null fields: {len(df_thresh)}")
````

## Enrichment
### Handling Sentinel Values Prior to Dropping
When external systems encode missing values with special values like `-999`:
```python
df['Score'] = df['Score'].replace(-999, np.nan)
df = df.dropna(subset=['Score'])
```

## Takeaways
- Inspect which columns contain missing values using `df.isna().sum()` before deciding on a strategy.
- Default `dropna()` (`how='any'`) is simple but can discard a significant percentage of observations.
- Use `subset=['ColName']` to restrict row dropping to mission-critical fields.
- Use `thresh=k` to enforce minimum acceptable data density per row.
- Always calculate and report the percentage of data lost when dropping rows.

## Conclusion
Dropping missing data should be done with deliberate intent. Targeted dropping via `subset` protects valuable data in secondary columns while removing invalid rows.
