## Overview
The default aesthetics of basic plotting libraries can look dated and unpolished. Professional reports, client presentations, and academic publications require **customised styling**, harmonious color palettes, clear typographic hierarchy, and **multi-panel subplots** that tell a cohesive visual story.

### Learning Objectives
- Apply built-in aesthetic themes (e.g. `seaborn-v0_8-whitegrid`, `fivethirtyeight`).
- Build multi-panel dashboards using `plt.subplots(nrows, ncols)`.
- Guide audience attention using `ax.annotate()` and custom arrow styles.
- Export publication-quality high-resolution figures (`dpi=300`, `bbox_inches='tight'`).

## 1. Setup: Academic Performance Dataset

We generate datasets tracking university campus enrolments, course pass rates, student attendance, and Weighted Average Marks (WAM).

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

campus_df = pd.DataFrame({
    'Campus': ['North Sydney', 'Melbourne', 'Brisbane', 'Strathfield'],
    'Enrolments': [4200, 4800, 2600, 1900],
    'Pass_Rate': [88.5, 85.0, 82.3, 91.0]
})

np.random.seed(42)
attendance = np.random.uniform(60, 100, size=150)
wam = np.clip(25 + 0.65 * attendance + np.random.normal(0, 5, size=150), 35, 100)
student_df = pd.DataFrame({'Attendance': attendance, 'WAM': wam})

print("Campus Summary:")
display(campus_df)
print("\nStudent Metrics Sample:")
display(student_df.head(5))
```

## 2. Choosing Built-In Matplotlib Styles

Matplotlib includes pre-configured style sheets that define fonts, background grids, and color cycles. Check available styles with `plt.style.available` and activate one with `plt.style.use('...')`.

```python
# Display available themes
print("Sample of available styles:")
print([s for s in plt.style.available if 'seaborn' in s or 'ggplot' in s or 'five' in s][:6])

# Set clean modern style
plt.style.use('seaborn-v0_8-whitegrid')
```

## 3. Building a 2x2 Multi-Panel Dashboard

Using `fig, axes = plt.subplots(2, 2, figsize=(12, 9))`, we assemble four complementary charts into a single cohesive report:
1. **Top-Left**: Campus Enrolments (Horizontal Bar Chart)
2. **Top-Right**: Attendance vs. WAM (Scatter with trendline and annotation)
3. **Bottom-Left**: Pass Rates (Vertical Bar Chart with threshold line)
4. **Bottom-Right**: WAM Distribution (Histogram with median line)

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9))
fig.suptitle('ACU Academic Performance & Engagement Dashboard', fontsize=15, fontweight='bold')

# 1. Campus Enrolments
ax1 = axes[0, 0]
bars1 = ax1.barh(campus_df['Campus'], campus_df['Enrolments'], color='#1f77b4', height=0.55)
ax1.bar_label(bars1, padding=4, fmt='%d')
ax1.set_title('Total Enrolments by Campus', fontsize=11, fontweight='bold')
ax1.set_xlabel('Students')
ax1.set_xlim(0, 5500)

# 2. Attendance vs WAM
ax2 = axes[0, 1]
scatter = ax2.scatter(student_df['Attendance'], student_df['WAM'], c=student_df['WAM'], cmap='viridis', alpha=0.7)
fig.colorbar(scatter, ax=ax2, label='WAM')
m, b = np.polyfit(student_df['Attendance'], student_df['WAM'], 1)
x_t = np.linspace(60, 100, 50)
ax2.plot(x_t, m * x_t + b, color='crimson', linestyle='--', linewidth=1.8, label='Linear Fit')
ax2.annotate('High Achiever', xy=(98, 92), xytext=(72, 95), arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5), fontweight='bold')
ax2.set_title('Attendance vs. WAM', fontsize=11, fontweight='bold')
ax2.set_xlabel('Attendance (%)')
ax2.set_ylabel('WAM')
ax2.legend(loc='lower right')

# 3. Pass Rates
ax3 = axes[1, 0]
pass_b = ax3.bar(campus_df['Campus'], campus_df['Pass_Rate'], color='#2ca02c', width=0.45)
ax3.bar_label(pass_b, padding=3, fmt='%.1f%%')
ax3.axhline(85.0, color='orange', linestyle=':', linewidth=1.8, label='Target (85%)')
ax3.set_title('Pass Rates by Campus', fontsize=11, fontweight='bold')
ax3.set_ylabel('Pass Rate (%)')
ax3.set_ylim(0, 100)
ax3.legend(loc='lower left')

# 4. WAM Distribution
ax4 = axes[1, 1]
ax4.hist(student_df['WAM'], bins=15, color='#9467bd', edgecolor='black', alpha=0.8)
med_wam = student_df['WAM'].median()
ax4.axvline(med_wam, color='red', linestyle='--', linewidth=2, label=f'Median: {med_wam:.1f}')
ax4.set_title('Cohort WAM Distribution', fontsize=11, fontweight='bold')
ax4.set_xlabel('WAM')
ax4.set_ylabel('Count')
ax4.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

## 4. Practical Exercises

### Exercise: 1x2 Executive Summary Dashboard
Below are unit enrolments and satisfaction ratings across four computing units.
1. Create a 1x2 figure (`plt.subplots(1, 2, figsize=(11, 4.5))`).
2. On `ax1`, plot a vertical bar chart of enrolments with bar labels.
3. On `ax2`, plot a bar chart of satisfaction ratings with a horizontal target line at 90%.

```python
units = ['ITEC102', 'ITEC105', 'ITEC108', 'ITEC110']
satisfaction = [92.5, 88.0, 84.5, 90.0]
enrolments = [320, 280, 240, 210]

# --- Student Code Here ---
# fig, (ax1, ax2) = plt.subplots(...)

# --- Solution ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4.5))

# Left: Enrolments
b1 = ax1.bar(units, enrolments, color='teal', width=0.5)
ax1.bar_label(b1, padding=3)
ax1.set_title('Unit Enrolments', fontsize=11, fontweight='bold')
ax1.set_ylabel('Students')

# Right: Satisfaction
b2 = ax2.bar(units, satisfaction, color='coral', width=0.5)
ax2.bar_label(b2, padding=3, fmt='%.1f%%')
ax2.axhline(90.0, color='grey', linestyle='--', label='Target (90%)')
ax2.set_title('Student Satisfaction Rating', fontsize=11, fontweight='bold')
ax2.set_ylabel('Satisfaction (%)')
ax2.set_ylim(0, 100)
ax2.legend()

plt.tight_layout()
plt.show()
```

## 5. Key Takeaways

1. **`plt.subplots(nrows, ncols)`**: The gold standard for multi-chart dashboard layouts.
2. **Styling Consistency**: Set a single global theme (`plt.style.use(...)`) at the beginning of your script or notebook.
3. **Direct Labelling & Target Lines**: Use `ax.bar_label()`, `ax.axhline()`, and `ax.annotate()` to communicate insights without forcing readers to guess.
4. **High Resolution**: Always specify `dpi=300` and `bbox_inches='tight'` when saving figures for papers or presentations.
