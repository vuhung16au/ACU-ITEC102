## Overview
A **Histogram** visualises the frequency distribution of a continuous numerical variable by partitioning the range of values into intervals called **bins** and counting how many observations fall into each bin.

### Learning Objectives
- Distinguish between bar charts (categorical) and histograms (continuous numerical data).
- Choose appropriate bin numbers (`bins=15`) or explicit domain-specific bin edges.
- Overlay central tendency reference lines (`ax.axvline()`) for mean and median.
- Identify skewness and multi-modal distributions.

## 1. Setup: Student Final Exam Marks

We generate synthetic exam marks (0 to 100) for a cohort of 250 introductory programming students.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

np.random.seed(42)
main_cohort = np.random.normal(loc=73, scale=11, size=220)
tail_cohort = np.random.normal(loc=42, scale=8, size=30)
exam_marks = np.clip(np.round(np.concatenate([main_cohort, tail_cohort]), 1), 0, 100)

print(f"Cohort size: {len(exam_marks)}")
print(f"Summary: Min={exam_marks.min()}, Mean={exam_marks.mean():.1f}, Median={np.median(exam_marks):.1f}, Max={exam_marks.max()}")
```

## 2. Basic Histogram with Mean and Median Lines

Using `ax.axvline()`, we overlay the **Mean** and **Median**. If the mean is lower than the median, the distribution is left-skewed (dragged down by struggling students).

```python
fig, ax = plt.subplots(figsize=(9, 5))

mean_val = np.mean(exam_marks)
median_val = np.median(exam_marks)

counts, bins, patches = ax.hist(
    exam_marks,
    bins=15,
    color='#1f77b4',
    edgecolor='black',
    alpha=0.75,
    label='Exam Marks'
)

# Add reference lines
ax.axvline(mean_val, color='crimson', linestyle='--', linewidth=2, label=f'Mean: {mean_val:.1f}')
ax.axvline(median_val, color='gold', linestyle='-.', linewidth=2, label=f'Median: {median_val:.1f}')
ax.axvline(50, color='darkgrey', linestyle=':', linewidth=1.5, label='Pass Mark (50)')

ax.set_title('ITEC102 Final Exam Mark Distribution (n=250)', fontsize=12, fontweight='bold')
ax.set_xlabel('Exam Mark (%)')
ax.set_ylabel('Number of Students')
ax.set_xlim(20, 100)
ax.grid(axis='y', linestyle=':', alpha=0.6)
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

## 3. Explicit Grade Boundary Bins

Instead of equal-width bins, pass an explicit list of bin edges: `bins=[0, 50, 65, 75, 85, 100]` matching standard Australian higher-education grading tiers.

```python
fig, ax = plt.subplots(figsize=(9, 5))

grade_bins = [0, 50, 65, 75, 85, 100]
grade_colors = ['#d9534f', '#f0ad4e', '#5bc0de', '#0275d8', '#5cb85c']

counts, bins, patches = ax.hist(exam_marks, bins=grade_bins, edgecolor='black', alpha=0.85)

# Color bins by grade tier
for patch, color in zip(patches, grade_colors):
    patch.set_facecolor(color)

ax.set_title('Exam Marks Binned by Australian Grade Tiers', fontsize=12, fontweight='bold')
ax.set_xlabel('Grade Tier')
ax.set_ylabel('Student Count')
ax.set_xticks(grade_bins)
ax.set_xticklabels(['0', '50\n(Fail)', '65\n(Pass)', '75\n(Credit)', '85\n(Dist)', '100\n(HD)'])
ax.grid(axis='y', linestyle=':', alpha=0.6)

# Add student count text above each tier
for count, left, right in zip(counts, grade_bins[:-1], grade_bins[1:]):
    ax.text((left + right)/2, count + 2, f'{int(count)} students', ha='center', fontsize=9, fontweight='bold')

ax.set_ylim(0, max(counts) + 12)
plt.tight_layout()
plt.show()
```

## 4. Practical Exercises

### Exercise: Visualising Skewness in House Prices
Economic and income data is typically **right-skewed** (a few very large values pull the mean far above the median).
1. Plot a histogram with 25 bins of the property prices dataset below.
2. Overlay vertical lines for the Median and Mean.
3. Observe how the right tail pulls the mean.

```python
np.random.seed(10)
prices = np.random.lognormal(mean=7.0, sigma=0.45, size=400)  # Prices in $k AUD

# --- Student Code Here ---
# fig, ax = plt.subplots(...)

# --- Solution ---
fig, ax = plt.subplots(figsize=(9, 5))
mean_p = np.mean(prices)
median_p = np.median(prices)

ax.hist(prices, bins=25, color='cadetblue', edgecolor='black', alpha=0.75)
ax.axvline(median_p, color='darkorange', linestyle='-', linewidth=2, label=f'Median: ${median_p:,.0f}k')
ax.axvline(mean_p, color='crimson', linestyle='--', linewidth=2, label=f'Mean: ${mean_p:,.0f}k')

ax.set_title('Sydney Property Price Distribution (Right Skewness)', fontsize=12, fontweight='bold')
ax.set_xlabel('Price ($k AUD)')
ax.set_ylabel('Property Count')
ax.grid(axis='y', linestyle=':', alpha=0.6)
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## 5. Key Takeaways

1. **Continuous Data Only**: Histograms bin continuous numbers; never use them for discrete categories (use bar charts instead).
2. **Bin Sensitivity**: Too few bins over-smooth the data; too many bins create jagged noise. Experiment with 10–30 bins.
3. **Mean vs. Median**: Symmetrical distributions have mean $\approx$ median. Right-skewed data has mean $>$ median; left-skewed data has mean $<$ median.
