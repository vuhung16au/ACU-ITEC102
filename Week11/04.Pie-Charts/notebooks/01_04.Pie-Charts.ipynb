## Overview
**Pie Charts** display proportional breakdowns of a whole (must sum to 100%). However, they are often misused or overused in reporting. Human visual perception judges lengths and positions on a common scale far more accurately than angles or wedge areas.

In this notebook, we cover:
1. **Constructing Pie Charts**: Setting `autopct`, colors, and `startangle`.
2. **Slice Emphasis**: Using `explode` to highlight key segments.
3. **Modern Donut Charts**: Using `wedgeprops=dict(width=...)` to create a hollow center for summary statistics.
4. **Visualisation Ethics & Guidelines**: When to use a pie chart, and why sorted bar charts are often superior.

## 1. Setup: Australian Electricity Generation Mix

We examine the breakdown of electricity generation in the Australian National Electricity Market (NEM).

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df_energy = pd.DataFrame({
    'Source': ['Coal', 'Solar', 'Wind', 'Gas', 'Hydro', 'Bioenergy/Storage'],
    'Share_pct': [46.5, 21.2, 14.8, 9.2, 6.5, 1.8]
})

print("Energy Generation Mix Dataset:")
display(df_energy)
```

## 2. Standard Pie Chart with Explode

Key parameters:
- `autopct='%1.1f%%'`: Formats slice percentage labels.
- `explode`: Offsets a specific slice outward for visual emphasis.
- `startangle`: Rotates the start of the pie (90° or 140° provides a balanced aesthetic).

```python
fig, ax = plt.subplots(figsize=(8, 6))

# Explode Solar slice outward
explode = (0, 0.1, 0, 0, 0, 0)
colors = ['#4a4a4a', '#ffbb33', '#33b5e5', '#ff4444', '#00C851', '#aa66cc']

wedges, texts, autotexts = ax.pie(
    df_energy['Share_pct'],
    explode=explode,
    labels=df_energy['Source'],
    autopct='%1.1f%%',
    pctdistance=0.75,
    startangle=140,
    colors=colors,
    wedgeprops=dict(edgecolor='white', linewidth=1.5)
)

for at in autotexts:
    at.set_color('black')
    at.set_fontweight('bold')

ax.set_title('Australian National Electricity Market Generation Mix', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Creating a Modern Donut Chart

A **Donut Chart** hollows out the center of the pie. The open ring feels less heavy visually and provides space for a prominent total or metric in the center.

```python
fig, ax = plt.subplots(figsize=(8, 6))

wedges, texts, autotexts = ax.pie(
    df_energy['Share_pct'],
    labels=df_energy['Source'],
    autopct='%1.1f%%',
    pctdistance=0.82,
    startangle=90,
    colors=colors,
    wedgeprops=dict(width=0.4, edgecolor='white', linewidth=2)  # width controls donut ring thickness
)

for at in autotexts:
    at.set_fontweight('bold')

# Center text
ax.text(0, 0, 'Total\n218 TWh', ha='center', va='center', fontsize=13, fontweight='bold', color='#333333')
ax.set_title('Australian Energy Mix (Donut Chart)', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

## 4. Visual Perception: Pie Chart vs. Horizontal Bar Chart

Notice how in the pie chart on the left, comparing Gas (9.2%) vs Hydro (6.5%) requires squinting at angles. In the horizontal bar chart on the right, the lengths provide an instant, unambiguous comparison.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Left: Pie Chart
ax1.pie(df_energy['Share_pct'], labels=df_energy['Source'], autopct='%1.1f%%', startangle=90)
ax1.set_title('Pie Chart (Harder to rank slices)', fontsize=11, fontweight='bold')

# Right: Sorted Horizontal Bar Chart
sorted_df = df_energy.sort_values(by='Share_pct', ascending=True)
bars = ax2.barh(sorted_df['Source'], sorted_df['Share_pct'], color='steelblue', height=0.6)
ax2.bar_label(bars, fmt='%.1f%%', padding=3, fontweight='bold')
ax2.set_xlim(0, 55)
ax2.set_xlabel('Share of Generation (%)')
ax2.set_title('Sorted Bar Chart (Instant, accurate ranking)', fontsize=11, fontweight='bold')
ax2.grid(axis='x', linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 5. Practical Exercises

### Exercise: University Cohort Composition Donut Chart
Below are university enrolments across residency and study mode.
1. Create a donut chart with `wedgeprops=dict(width=0.45)`.
2. Explode the `International Full-Time` slice outward by `0.1`.
3. Add the total enrolment count in the center of the donut.

```python
cohort = pd.DataFrame({
    'Cohort': ['Domestic Full-Time', 'Domestic Part-Time', 'International Full-Time', 'International Part-Time'],
    'Students': [5800, 2200, 1500, 500]
})

# --- Student Code Here ---
# fig, ax = plt.subplots(...)

# --- Solution ---
fig, ax = plt.subplots(figsize=(7, 6))
explode = (0, 0, 0.1, 0)
palette = ['#1f77b4', '#aec7e8', '#ff7f0e', '#ffbb78']

ax.pie(
    cohort['Students'],
    explode=explode,
    labels=cohort['Cohort'],
    autopct='%1.1f%%',
    pctdistance=0.8,
    colors=palette,
    wedgeprops=dict(width=0.45, edgecolor='white', linewidth=2)
)

total = cohort['Students'].sum()
ax.text(0, 0, f'Enrolments\n{total:,}', ha='center', va='center', fontsize=11, fontweight='bold')
ax.set_title('ACU Student Enrolment Cohorts', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 6. Key Takeaways

1. **Rule of 5**: Never use pie charts with more than 5–6 slices; group minor categories into an 'Other' bucket.
2. **Must Sum to 100%**: Pie charts are strictly for mutually exclusive percentages that sum to 100%.
3. **When to Choose Bar Charts**: If accurate numerical comparison or ranking is required, always prefer a sorted bar chart over a pie chart.
