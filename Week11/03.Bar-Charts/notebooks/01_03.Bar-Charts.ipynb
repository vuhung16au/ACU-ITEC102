## Overview
**Bar Charts** are the most versatile and widely understood visualization for comparing values across discrete categories. Unlike line plots (which imply continuous flow), bar charts isolate individual cohorts, making rankings and volume comparisons immediate.

### Learning Objectives
- Create vertical bar charts (`ax.bar()`) and horizontal bar charts (`ax.barh()`).
- Add direct data labels using `ax.bar_label()`.
- Construct grouped (clustered) bar charts using bar-width coordinate offsets.
- Build stacked bar charts using the `bottom` argument to show part-to-whole relationships.

## 1. Setup: Faculty Enrolment Data

We compare student enrolments across university faculties and Australian campuses.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df_enrolments = pd.DataFrame({
    'Faculty': [
        'Faculty of Health Sciences',
        'Faculty of Education & Arts',
        'Faculty of Law & Business',
        'Faculty of Theology & Philosophy'
    ],
    'Sydney': [3200, 2100, 2400, 850],
    'Melbourne': [3500, 2300, 1900, 920],
    'Brisbane': [2100, 1400, 1600, 610]
})

print("Faculty Enrolments Dataset:")
display(df_enrolments)
```

## 2. Vertical Bar Chart with Data Labels

Using `ax.bar()` creates vertical columns. Matplotlib's `ax.bar_label()` automatically places numeric values directly above each bar, eliminating guesswork.

```python
fig, ax = plt.subplots(figsize=(8, 4.5))

short_names = ['Health', 'Education', 'Business', 'Theology']
bars = ax.bar(short_names, df_enrolments['Sydney'], color='#1f77b4', width=0.5, edgecolor='black')

# Directly annotate bars with exact numbers
ax.bar_label(bars, padding=4, fontsize=10, fontweight='bold')

ax.set_title('Sydney Enrolments by Faculty', fontsize=12, fontweight='bold')
ax.set_ylabel('Student Count')
ax.set_ylim(0, 4000)
ax.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 3. Horizontal Bar Charts for Long Labels

When category names are descriptive or long, vertical charts force ugly 45° or 90° rotated labels. **Horizontal bar charts (`ax.barh()`)** display labels horizontally, making them much easier to read.

```python
fig, ax = plt.subplots(figsize=(9, 4.5))

y_pos = np.arange(len(df_enrolments['Faculty']))
bars = ax.barh(y_pos, df_enrolments['Sydney'], color='#2ca02c', height=0.55, edgecolor='black')

ax.set_yticks(y_pos, df_enrolments['Faculty'])
ax.bar_label(bars, padding=5, fontsize=9)
ax.set_title('Enrolments by Faculty (Horizontal Layout)', fontsize=12, fontweight='bold')
ax.set_xlabel('Student Count')
ax.set_xlim(0, 4000)
ax.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 4. Grouped (Clustered) Bar Charts

To compare multiple series side-by-side (e.g. Sydney, Melbourne, Brisbane), we offset the x-coordinates by bar width (`x - width`, `x`, `x + width`).

```python
fig, ax = plt.subplots(figsize=(10, 5))

x = np.arange(len(df_enrolments['Faculty']))
width = 0.25

ax.bar(x - width, df_enrolments['Sydney'], width, label='Sydney', color='#004c6d')
ax.bar(x, df_enrolments['Melbourne'], width, label='Melbourne', color='#257d9d')
ax.bar(x + width, df_enrolments['Brisbane'], width, label='Brisbane', color='#de425b')

ax.set_xticks(x, short_names)
ax.set_title('Cross-Campus Faculty Enrolments Comparison', fontsize=13, fontweight='bold')
ax.set_ylabel('Student Count')
ax.set_ylim(0, 4200)
ax.grid(axis='y', linestyle=':', alpha=0.6)
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## 5. Stacked Bar Charts

Stacked bar charts stack sub-categories vertically using the `bottom` argument, displaying both individual breakdown and total volume.

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(short_names, df_enrolments['Sydney'], label='Sydney', color='#1f77b4', width=0.5)
ax.bar(short_names, df_enrolments['Melbourne'], bottom=df_enrolments['Sydney'], label='Melbourne', color='#ff7f0e', width=0.5)

bottom_two = df_enrolments['Sydney'] + df_enrolments['Melbourne']
ax.bar(short_names, df_enrolments['Brisbane'], bottom=bottom_two, label='Brisbane', color='#2ca02c', width=0.5)

ax.set_title('Total Faculty Enrolments (Stacked by Campus)', fontsize=12, fontweight='bold')
ax.set_ylabel('Total Enrolments')
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## 6. Practical Exercises

### Exercise: Comparing Tech Graduate Starting Salaries
Below are starting salaries ($AUD) for Bachelor vs. Master graduates in computing.
1. Create a grouped horizontal bar chart comparing Bachelor vs Master salaries across disciplines.
2. Add a clear legend and gridlines.

```python
salaries = pd.DataFrame({
    'Discipline': ['Cloud Arch', 'Cyber Security', 'Data Science', 'Software Eng'],
    'Bachelor': [92000, 88000, 85000, 82000],
    'Master': [115000, 108000, 105000, 102000]
})

# --- Student Code Here ---
# fig, ax = plt.subplots(...)

# --- Solution ---
fig, ax = plt.subplots(figsize=(9, 5))
y = np.arange(len(salaries['Discipline']))
h = 0.35

ax.barh(y - h/2, salaries['Bachelor'], h, label='Bachelor', color='steelblue')
ax.barh(y + h/2, salaries['Master'], h, label='Master', color='darkorange')

ax.set_yticks(y, salaries['Discipline'])
ax.set_title('Tech Starting Salaries by Degree Level ($AUD)', fontsize=12, fontweight='bold')
ax.set_xlabel('Salary ($AUD)')
ax.set_xlim(0, 130000)
ax.grid(axis='x', linestyle=':', alpha=0.6)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

## 7. Key Takeaways

1. **Vertical vs Horizontal**: Always prefer horizontal bars (`ax.barh()`) when category labels are long.
2. **Grouped Bars**: Use coordinate offsets (`x ± width/2`) to place bars side-by-side.
3. **Stacked Bars**: Use `bottom=...` to display component breakdowns and cumulative totals.
4. **Direct Labelling**: Use `ax.bar_label()` to present exact values above bars without cluttering the y-axis.
