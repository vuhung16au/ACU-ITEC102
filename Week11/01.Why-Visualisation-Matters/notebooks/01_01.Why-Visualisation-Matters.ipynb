## Overview
It is tempting to rely solely on summary statistics like the mean, median, standard deviation, and correlation to understand data. However, **summary statistics alone can dangerously mislead**.

In 1973, English statistician Francis Anscombe created four synthetic datasets—now known as **Anscombe's Quartet**—to illustrate why visualising data is essential before conducting analysis or building models.

### Learning Objectives
- Understand why datasets with identical statistical summaries can have completely different structures.
- Master Matplotlib's Object-Oriented interface (`Figure` and `Axes`).
- Detect non-linear trends, clustering, and high-leverage outliers through visual inspection.

## 1. The Statistical Illusion

Below are four distinct datasets. Let's inspect their summary statistics.

````python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Anscombe's Quartet Datasets
x1 = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5], dtype=float)
y1 = np.array([8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68])

x2 = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5], dtype=float)
y2 = np.array([9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74])

x3 = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5], dtype=float)
y3 = np.array([7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73])

x4 = np.array([8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8], dtype=float)
y4 = np.array([6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89])

quartet = [(x1, y1), (x2, y2), (x3, y3), (x4, y4)]

# Compute statistical summaries
rows = []
for i, (x, y) in enumerate(quartet, 1):
    slope, intercept = np.polyfit(x, y, deg=1)
    rows.append({
        'Dataset': f'Dataset {i}',
        'x Mean': round(float(np.mean(x)), 2),
        'x Variance': round(float(np.var(x, ddof=1)), 2),
        'y Mean': round(float(np.mean(y)), 2),
        'y Variance': round(float(np.var(y, ddof=1)), 2),
        'Correlation': round(float(np.corrcoef(x, y)[0, 1]), 3),
        'Linear Fit': f'y = {slope:.2f}x + {intercept:.2f}'
    })

display(pd.DataFrame(rows))
````

## 2. Visualising the Quartet

When we plot each dataset, we discover four completely distinct patterns:
- **Dataset I**: Standard linear relationship with moderate scatter.
- **Dataset II**: Clear quadratic (non-linear) curve where a linear model is invalid.
- **Dataset III**: Perfect linear line with a single vertical outlier pulling the regression.
- **Dataset IV**: Pure vertical cluster with a single high-leverage point creating an artificial correlation.

````python
# Plot all 4 datasets in a 2x2 grid using Matplotlib's Object-Oriented interface
fig, axes = plt.subplots(2, 2, figsize=(10, 8), sharex=True, sharey=True)
fig.suptitle("Anscombe's Quartet: Identical Stats, Radically Different Shapes", fontsize=14, fontweight='bold')

titles = [
    "Dataset I: Standard Linear",
    "Dataset II: Non-Linear (Quadratic)",
    "Dataset III: Linear with Outlier",
    "Dataset IV: Clustered with Leverage Point"
]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

x_fit = np.linspace(2, 20, 100)
y_fit = 3.0 + 0.5 * x_fit

for idx, (ax, (x, y)) in enumerate(zip(axes.flat, quartet)):
    ax.scatter(x, y, color=colors[idx], s=70, edgecolors='black', label='Observed Data', zorder=3)
    ax.plot(x_fit, y_fit, color='black', linestyle='--', linewidth=1.5, label='Fitted Line (y=0.5x+3.0)')
    ax.set_title(titles[idx], fontsize=11)
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='upper left', fontsize=9)
    ax.set_xlim(2, 20)
    ax.set_ylim(2, 14)

plt.tight_layout()
plt.show()
````

## 3. Matplotlib Anatomy: Figure vs. Axes

In Matplotlib:
- **`Figure`**: The overarching window or canvas holding all chart elements.
- **`Axes`**: An individual plot/subplot region with its own coordinate system, x/y axes, title, and data lines.

The modern, professional way to code Matplotlib is the **Object-Oriented (OO) interface**:
```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(...)
ax.set_title(...)
```

## 4. Practical Exercises

### Exercise 1: Identifying Outlier Distortion
Below are weekly study hours and final marks for 12 students. A data entry error recorded one student as having studied 50 hours with a mark of 15.
1. Compute the correlation coefficient with the outlier included.
2. Filter out the outlier (`study_hours <= 20`) and recalculate the correlation.

````python
study_hours = np.array([2, 3, 4, 5, 6, 7, 8, 9, 10, 12, 14, 50])
exam_marks = np.array([45, 52, 58, 63, 67, 72, 75, 80, 84, 88, 92, 15])

# --- Student Code Here ---
# corr_raw = ...
# corr_clean = ...

# --- Solution ---
corr_raw = np.corrcoef(study_hours, exam_marks)[0, 1]
clean_mask = study_hours <= 20
corr_clean = np.corrcoef(study_hours[clean_mask], exam_marks[clean_mask])[0, 1]

print(f"Raw Correlation (with outlier):   {corr_raw:.3f}")
print(f"Clean Correlation (true trend):    {corr_clean:.3f}")
````

### Exercise 2: Visualising the Outlier
Plot a scatter plot highlighting standard observations in teal and the outlier in red.

````python
# --- Student Code Here ---
# fig, ax = plt.subplots(...)

# --- Solution ---
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(study_hours[clean_mask], exam_marks[clean_mask], color='teal', s=70, label='Standard Students')
ax.scatter(study_hours[~clean_mask], exam_marks[~clean_mask], color='crimson', s=120, marker='X', label='Identified Outlier')
ax.set_title('Study Hours vs. Exam Performance: Unmasking Data Errors', fontsize=12)
ax.set_xlabel('Weekly Study Hours')
ax.set_ylabel('Exam Mark (%)')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend()
plt.show()
````

## 5. Key Takeaways

1. **Never trust summary tables blindly**: Outliers and non-linear shapes can hide behind identical averages.
2. **Always plot your raw data first**: Visualisation is the first and most crucial step of Exploratory Data Analysis (EDA).
3. **Figure vs. Axes**: Always use `fig, ax = plt.subplots()` for granular control and cleaner code.
