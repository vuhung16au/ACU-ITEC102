## Overview
**Line Plots** are the standard visual tool for displaying continuous data, particularly **time series** (data points ordered chronologically). They help viewers spot trends, seasonal fluctuations, turning points, and outliers across time.

### Learning Objectives
- Plot single and multiple continuous trends on an `Axes`.
- Customise line aesthetics (`color`, `linestyle`, `linewidth`, `marker`).
- Add clear axis labels, titles, gridlines, and legends.
- Annotate key peaks and milestones directly on the chart.

## 1. Setup: Australian Capital City Temperatures

We record average monthly maximum temperatures (°C) across Sydney, Melbourne, and Brisbane.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
df_temps = pd.DataFrame({
    'Month': months,
    'Sydney': [26.0, 25.8, 24.8, 22.5, 19.5, 17.0, 16.4, 17.9, 20.1, 22.2, 23.7, 25.2],
    'Melbourne': [26.0, 25.8, 23.9, 20.3, 16.7, 14.1, 13.5, 15.0, 17.3, 19.7, 22.0, 24.2],
    'Brisbane': [30.4, 30.1, 29.0, 27.2, 24.5, 22.0, 21.9, 23.4, 25.7, 27.1, 28.3, 29.6]
})

print("Monthly Temperature Dataset (°C):")
display(df_temps)
```

## 2. Basic Single Line Plot

Using the Object-Oriented interface (`fig, ax = plt.subplots()`):
- `ax.plot(x, y)`: Draws the connected line.
- `marker='o'`: Highlights individual observation points.
- `grid(True)`: Adds reference gridlines for easier reading.

```python
fig, ax = plt.subplots(figsize=(8, 4.5))

ax.plot(
    df_temps['Month'],
    df_temps['Sydney'],
    color='#1f77b4',
    linestyle='-',
    linewidth=2.0,
    marker='o',
    markersize=6,
    label='Sydney'
)

ax.set_title('Sydney Monthly Mean Maximum Temperature', fontsize=12, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Temperature (°C)')
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

## 3. Comparative Multi-Line Plot

Plot multiple series on the same Axes to allow instant visual comparison. Distinguish lines using different colors, line styles (`'-'`, `'--'`, `'-.'`), and markers (`'o'`, `'s'`, `'^'`).

```python
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(df_temps['Month'], df_temps['Sydney'], color='#004c6d', linestyle='-', linewidth=2.2, marker='o', label='Sydney')
ax.plot(df_temps['Month'], df_temps['Melbourne'], color='#257d9d', linestyle='--', linewidth=2.0, marker='s', label='Melbourne')
ax.plot(df_temps['Month'], df_temps['Brisbane'], color='#de425b', linestyle='-.', linewidth=2.0, marker='^', label='Brisbane')

ax.set_title('Australian Capital Cities: Annual Temperature Profiles', fontsize=13, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Mean Maximum Temperature (°C)')
ax.set_ylim(10, 35)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower center', ncol=3, frameon=True)

plt.tight_layout()
plt.show()
```

## 4. Practical Exercises

### Exercise: Australian Renewable Energy Generation Trends
Below is monthly electricity generation (GWh) from Solar and Wind in Australia.
1. Plot both `Solar_GWh` and `Wind_GWh` across months on a single figure.
2. Style Solar with a golden line and circle markers; style Wind with a teal dashed line and triangle markers.
3. Use `ax.annotate()` to highlight the peak solar generation month.

```python
energy = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Solar_GWh': [3200, 2900, 2600, 2100, 1700, 1450, 1600, 2050, 2500, 3100, 3400, 3600],
    'Wind_GWh': [2400, 2200, 2500, 2700, 3100, 3400, 3300, 3200, 2900, 2800, 2600, 2500]
})

# --- Student Code Here ---
# fig, ax = plt.subplots(...)

# --- Solution ---
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(energy['Month'], energy['Solar_GWh'], color='goldenrod', linestyle='-', linewidth=2.2, marker='o', label='Solar Generation')
ax.plot(energy['Month'], energy['Wind_GWh'], color='teal', linestyle='--', linewidth=2.0, marker='^', label='Wind Generation')

# Annotate summer peak solar
peak_val = energy['Solar_GWh'].max()
peak_month = energy.loc[energy['Solar_GWh'].idxmax(), 'Month']
ax.annotate(
    f'Peak Solar: {peak_val} GWh',
    xy=(peak_month, peak_val),
    xytext=(8, 3800),
    arrowprops=dict(facecolor='black', shrink=0.08, width=1, headwidth=6),
    fontweight='bold'
)

ax.set_title('Australian Renewable Energy Generation (GWh)', fontsize=12, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Generation (GWh)')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower left')
plt.tight_layout()
plt.show()
```

## 5. Key Takeaways

1. **Continuous Data**: Line plots are ideal for continuous sequences where the ordering between points matters (especially time).
2. **Styling Distinction**: When plotting multiple series, vary line styles and markers in addition to color to ensure readability for color-blind viewers or monochrome printouts.
3. **Annotations**: Use `ax.annotate()` to guide viewer attention directly to anomalies or peaks.
